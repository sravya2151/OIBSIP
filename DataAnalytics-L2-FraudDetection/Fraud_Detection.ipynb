## Why Accuracy Is Misleading in Fraud Detection

The credit card fraud dataset is highly imbalanced because fraudulent transactions represent only about 0.17% of all transactions.

In such an imbalanced dataset, accuracy can be misleading. For example, a model that predicts every transaction as legitimate could achieve approximately 99.83% accuracy while failing to detect any fraudulent transactions.

Therefore, accuracy alone is not a suitable evaluation metric for fraud detection. Precision, Recall, F1-Score, and AUC-ROC provide a more meaningful evaluation of model performance.

Recall is particularly important when missing a fraudulent transaction is more costly than investigating a legitimate transaction incorrectly flagged as fraud.

```python
# Separate features and target variable

X = df.drop(columns=["Class", "Log_Amount", "Hour"])
y = df["Class"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nFraud cases in training data:")
print(y_train.value_counts())

print("\nFraud cases in testing data:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(pd.Series(y_train_smote).value_counts())
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(pd.Series(y_train_smote).value_counts())
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_smote, y_train_smote)

print("Logistic Regression model trained successfully!")
```

```python
y_pred_logistic = logistic_model.predict(X_test_scaled)

y_prob_logistic = logistic_model.predict_proba(X_test_scaled)[:, 1]

print("Predictions completed!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

logistic_accuracy = accuracy_score(y_test, y_pred_logistic)
logistic_precision = precision_score(y_test, y_pred_logistic)
logistic_recall = recall_score(y_test, y_pred_logistic)
logistic_f1 = f1_score(y_test, y_pred_logistic)
logistic_auc = roc_auc_score(y_test, y_prob_logistic)

print("Logistic Regression Results")
print("---------------------------")
print("Accuracy :", round(logistic_accuracy, 4))
print("Precision:", round(logistic_precision, 4))
print("Recall   :", round(logistic_recall, 4))
print("F1-Score :", round(logistic_f1, 4))
print("ROC-AUC  :", round(logistic_auc, 4))
```

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(X_train_smote, y_train_smote)

print("Random Forest model trained successfully!")
```

```python
y_pred_rf = random_forest_model.predict(X_test_scaled)

y_prob_rf = random_forest_model.predict_proba(X_test_scaled)[:, 1]

print("Random Forest predictions completed!")
```

```python
rf_accuracy = accuracy_score(y_test, y_pred_rf)
rf_precision = precision_score(y_test, y_pred_rf)
rf_recall = recall_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)
rf_auc = roc_auc_score(y_test, y_prob_rf)

print("Random Forest Results")
print("--------------------")
print("Accuracy :", round(rf_accuracy, 4))
print("Precision:", round(rf_precision, 4))
print("Recall   :", round(rf_recall, 4))
print("F1-Score :", round(rf_f1, 4))
print("ROC-AUC  :", round(rf_auc, 4))
```

```python
from sklearn.metrics import confusion_matrix

cm_logistic = confusion_matrix(y_test, y_pred_logistic)
cm_rf = confusion_matrix(y_test, y_pred_rf)

print("Logistic Regression Confusion Matrix:")
print(cm_logistic)

print("\nRandom Forest Confusion Matrix:")
print(cm_rf)
```

```python
from sklearn.metrics import roc_curve, auc

# Calculate ROC curve values
fpr_logistic, tpr_logistic, _ = roc_curve(y_test, y_prob_logistic)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_prob_rf)

# Calculate AUC
auc_logistic = auc(fpr_logistic, tpr_logistic)
auc_rf = auc(fpr_rf, tpr_rf)

# Plot ROC curves
plt.figure(figsize=(8, 6))

plt.plot(
    fpr_logistic,
    tpr_logistic,
    label=f"Logistic Regression (AUC = {auc_logistic:.4f})"
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label=f"Random Forest (AUC = {auc_rf:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title("ROC-AUC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.savefig("images/roc_curve.png", bbox_inches="tight")
plt.show()
```

## ROC-AUC Analysis

The ROC-AUC curve evaluates how well the models distinguish fraudulent transactions from legitimate transactions across different classification thresholds.

Logistic Regression achieved a ROC-AUC of approximately 0.9708, while Random Forest achieved approximately 0.9685. Both models demonstrate strong ability to distinguish between the two classes.

Although Logistic Regression has a slightly higher ROC-AUC, Random Forest provides a much better balance of precision, recall, and F1-score. Therefore, Random Forest is selected as the better overall model for this project.

## Scalability: Handling 1 Million Transactions per Hour

A production fraud detection system may need to process around 1 million transactions per hour. The current Python and scikit-learn pipeline demonstrates the machine learning approach, but a production system would require scalable infrastructure.

### Key scalability considerations

- **Real-time processing:** Use a streaming platform such as Apache Kafka to receive transactions continuously.
- **Distributed processing:** Technologies such as Apache Spark or Apache Flink can process large transaction volumes in parallel.
- **Model serving:** Deploy the trained model as a REST API or model-serving service so applications can request fraud predictions in real time.
- **Feature engineering:** Build efficient pipelines to calculate transaction features with low latency.
- **Parallel processing:** Multiple service instances can process transactions simultaneously to handle high traffic.
- **Monitoring:** Monitor prediction latency, fraud detection performance, false positives, and changes in data patterns.
- **Model retraining:** Retrain the model periodically using newly labeled transactions to handle changing fraud patterns.
- **Threshold tuning:** Adjust the classification threshold according to the business cost of missed fraud versus false alarms.

### Conclusion

For 1 million transactions per hour, the model should be deployed as part of a distributed, real-time fraud detection architecture rather than running as a single local Python script. Horizontal scaling, efficient feature processing, model monitoring, and periodic retraining would help maintain reliable performance as transaction volume increases.

# Conclusion and Business Recommendations

## Conclusion

This project developed a machine learning pipeline for detecting fraudulent financial transactions using the Credit Card Fraud Detection dataset.

The dataset was highly imbalanced, with fraudulent transactions representing only about 0.17% of all transactions. Because of this severe class imbalance, accuracy alone was not considered a reliable evaluation metric.

Exploratory Data Analysis was performed to examine transaction amounts, class distribution, and transaction patterns across the approximate time of day.

SMOTE (Synthetic Minority Oversampling Technique) was applied only to the training data to address class imbalance while keeping the test data representative of the original distribution.

Two machine learning models were trained and evaluated:

- Logistic Regression
- Random Forest

Random Forest achieved the best overall balance of precision, recall, and F1-score, with a precision of 0.8710, recall of 0.8265, F1-score of 0.8482, and ROC-AUC of 0.9685.

Logistic Regression achieved a higher recall of 0.9184, meaning it detected a larger proportion of fraudulent transactions, but its precision was only 0.0578, resulting in many false-positive alerts.

Therefore, Random Forest was selected as the preferred overall model because it provides a much better balance between detecting fraud and reducing false alarms.

## Business Recommendations

1. **Prioritize fraud recall:** Missing a fraudulent transaction can cause significant financial loss, so recall should be closely monitored.

2. **Control false positives:** Precision should also be monitored because excessive false alerts can inconvenience legitimate customers and increase investigation costs.

3. **Use threshold tuning:** The prediction threshold can be adjusted according to the business cost of missed fraud versus false alarms.

4. **Monitor model performance:** Fraud patterns can change over time, so performance and data drift should be continuously monitored.

5. **Scale the system:** For very high transaction volumes, such as 1 million transactions per hour, the model should be deployed using scalable real-time and distributed processing infrastructure.

6. **Retrain periodically:** New transaction data and confirmed fraud cases should be used to periodically retrain and improve the model.
