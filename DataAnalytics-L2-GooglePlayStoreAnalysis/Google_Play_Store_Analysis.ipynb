```python
print("Rows before removing duplicates:", len(df))

df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

```python
df["Installs"] = (
    df["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

print(df["Installs"].head())
print("\nInstalls data type:", df["Installs"].dtype)
```

```python
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

print(df["Price"].head())
print("\nPrice data type:", df["Price"].dtype)
```

```python
def convert_size(size):
    if pd.isna(size):
        return np.nan
    
    size = str(size).strip().replace(",", "")
    
    if size.endswith("M"):
        return float(size[:-1])
    
    elif size.endswith("k"):
        return float(size[:-1]) / 1024
    
    elif size.endswith("G"):
        return float(size[:-1]) * 1024
    
    else:
        return np.nan

df["Size_MB"] = df["Size"].apply(convert_size)

print(df[["Size", "Size_MB"]].head(10))
```

```python
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")

print("Rating data type:", df["Rating"].dtype)
print("\nMissing ratings:", df["Rating"].isnull().sum())
```

```python
print("Dataset shape after cleaning:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Fill missing numeric values with the median
df["Rating"] = df["Rating"].fillna(df["Rating"].median())
df["Rating Count"] = df["Rating Count"].fillna(0)
df["Installs"] = df["Installs"].fillna(0)
df["Minimum Installs"] = df["Minimum Installs"].fillna(0)
df["Size_MB"] = df["Size_MB"].fillna(df["Size_MB"].median())

# Fill missing text values with "Unknown"
text_columns = [
    "App Name",
    "Currency",
    "Minimum Android",
    "Developer Id",
    "Developer Website",
    "Developer Email",
    "Released",
    "Privacy Policy"
]

for column in text_columns:
    df[column] = df[column].fillna("Unknown")

print("Missing values handled successfully!")
print("Total remaining missing values:", df.isnull().sum().sum())
```

```python
print("Final dataset shape:", df.shape)
print("\nTotal missing values:", df.isnull().sum().sum())
print("Total duplicate rows:", df.duplicated().sum())
```

```python
df.to_csv("Google-Playstore-Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
from textblob import TextBlob

print("TextBlob imported successfully!")
```

```python
test_review = reviews["Translated_Review"].iloc[0]

analysis = TextBlob(test_review)

print("Review:")
print(test_review)

print("\nPolarity:", analysis.sentiment.polarity)
print("Subjectivity:", analysis.sentiment.subjectivity)
```

```python
def get_polarity(text):
    return TextBlob(text).sentiment.polarity

reviews["Sentiment_Polarity"] = reviews["Translated_Review"].apply(get_polarity)

print("Sentiment polarity calculated successfully!")
print(reviews[["Translated_Review", "Sentiment_Polarity"]].head(10))
```

```python
def classify_sentiment(polarity):
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

reviews["Sentiment_Label"] = reviews["Sentiment_Polarity"].apply(
    classify_sentiment
)

print("Sentiment classification completed!")
print(reviews["Sentiment_Label"].value_counts())
```

```python
sentiment_counts = reviews["Sentiment_Label"].value_counts()

sentiment_percentages = (
    reviews["Sentiment_Label"]
    .value_counts(normalize=True) * 100
)

print("Sentiment counts:")
print(sentiment_counts)

print("\nSentiment percentages:")
print(sentiment_percentages.round(2))
```

```python
# Clean app names in both datasets for matching
reviews["App_Clean"] = (
    reviews["App"]
    .astype(str)
    .str.strip()
    .str.lower()
)

app_categories = (
    df[["App Name", "Category"]]
    .dropna(subset=["App Name"])
    .assign(
        App_Clean=lambda x: (
            x["App Name"]
            .astype(str)
            .str.strip()
            .str.lower()
        )
    )
    [["App_Clean", "Category"]]
    .drop_duplicates("App_Clean")
)

# Merge reviews with app categories
reviews_with_category = reviews.merge(
    app_categories,
    on="App_Clean",
    how="left"
)

print("Reviews matched to categories:",
      reviews_with_category["Category"].notna().sum())

print("Reviews without category:",
      reviews_with_category["Category"].isna().sum())
```

```python
category_sentiment = (
    reviews_with_category
    .dropna(subset=["Category"])
    .groupby("Category")["Sentiment_Polarity"]
    .agg(["mean", "count"])
    .sort_values("mean", ascending=False)
)

print("Top 10 most positive categories:")
print(category_sentiment.head(10).round(3))

print("\nTop 10 most negative categories:")
print(category_sentiment.tail(10).round(3))
```

```python
top_positive = category_sentiment.head(10)
top_negative = category_sentiment.tail(10)

sentiment_chart_data = pd.concat([
    top_positive,
    top_negative
]).sort_values("mean")

plt.figure(figsize=(12, 9))

plt.barh(
    sentiment_chart_data.index,
    sentiment_chart_data["mean"]
)

plt.title("Average Sentiment Polarity by App Category")
plt.xlabel("Average Sentiment Polarity")
plt.ylabel("Category")

plt.axvline(0, linewidth=1)

plt.tight_layout()

plt.savefig(
    "images/sentiment_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Conclusion & Business Recommendations

The Google Play Store analysis provides several useful insights for app developers:

### 1. Focus on differentiation in saturated categories

The **Education** category contains approximately **241,090 apps**, making it the largest category in the dataset. Developers entering highly populated categories should identify specific user needs and offer unique features to stand out from competitors.

### 2. Free and freemium models dominate the market

Approximately **98.05% of the apps are free**, while only **1.95% are paid**. Among paid apps, the median price is approximately **$1.99**. This suggests that developers should consider free or freemium strategies to reach a larger user base, with optional premium features or in-app monetization.

### 3. App size alone does not drive installs

The correlation between **app size and installs is approximately 0.0063**, indicating almost no linear relationship. Developers should therefore focus more on app quality, useful features, user experience, performance, and engagement rather than relying on app size as a factor for increasing downloads.

### Additional Insights

* The **Arcade** category generated the highest estimated revenue from paid apps, followed by **Action** and **Tools**.
* **Role Playing** had the highest average app rating among categories at approximately **3.37**.
* **Events** had the lowest average app rating at approximately **1.29**, indicating an opportunity to investigate user experience issues in this category.
* User sentiment was generally positive across the matched review categories. **Comics** had the highest average sentiment polarity, although it was based on a relatively small number of reviews.
* Categories such as **Strategy** and **Video Players & Editors** had the lowest average sentiment polarity among the analyzed categories, suggesting areas where developers could investigate user feedback and improve their products.

### Final Recommendation

Developers should prioritize **market differentiation, strong user experience, and sustainable monetization strategies**. Combining category competition, ratings, installs, pricing, and user sentiment can help identify promising opportunities and guide future app development decisions.
