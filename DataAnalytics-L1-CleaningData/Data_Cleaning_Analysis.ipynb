## Missing Data Handling

The columns `Peak` and `All Time Peak` contain missing values.

Since these columns represent chart ranking information, the missing values cannot be reliably calculated from other columns. Mean or median imputation would also be inappropriate because ranking values are not continuous measurements.

Therefore, missing values in these columns are replaced with the value `Unknown`. This preserves the remaining information about the tours while clearly indicating that the ranking information was unavailable.

No rows are deleted because the dataset contains only 20 records and the other fields contain useful information.

```python
df['Peak'] = df['Peak'].fillna('Unknown')
df['All Time Peak'] = df['All Time Peak'].fillna('Unknown')

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
def clean_rank(value):
    if value == 'Unknown':
        return value
    
    value = str(value)
    return value.split('[')[0]

df['Peak'] = df['Peak'].apply(clean_rank)
df['All Time Peak'] = df['All Time Peak'].apply(clean_rank)

print("Peak values:")
print(df['Peak'].unique())

print("\nAll Time Peak values:")
print(df['All Time Peak'].unique())
```

```python
money_columns = [
    'Actual\xa0gross',
    'Adjusted\xa0gross (in 2022 dollars)',
    'Average gross'
]

for column in money_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.replace(r'[$,\[\]a-zA-Z]', '', regex=True)
        .astype(float)
    )

print(df[money_columns].dtypes)
```

```python
df['Tour title'] = (
    df['Tour title']
    .str.replace(r'\[[^\]]*\]', '', regex=True)
    .str.replace('†', '', regex=False)
    .str.replace('‡', '', regex=False)
    .str.replace('*', '', regex=False)
    .str.strip()
)

print(df['Tour title'].head())
```

```python
df.drop(columns=['Ref.'], inplace=True)

print("Columns after removing Ref.:")
print(df.columns.tolist())
```

## Standardisation

The dataset contained inconsistent formatting caused by web-scraping from Wikipedia.

Ranking columns contained citation markers such as `[4]` and `[7]`, which were removed while preserving the actual ranking values.

Monetary columns contained dollar signs, commas, and citation letters such as `[b]` and `[e]`. These non-numeric characters were removed and the values were converted to `float` for numerical analysis.

Tour titles contained Wikipedia reference symbols such as `†`, `‡`, `*`, and citation markers. These were removed to retain clean tour names.

The `Ref.` column contained only source citation references and was not required for analysis, so it was removed.

```python
df['Peak'] = pd.to_numeric(df['Peak'], errors='coerce')
df['All Time Peak'] = pd.to_numeric(df['All Time Peak'], errors='coerce')

print("Ranking data types:")
print(df[['Peak', 'All Time Peak']].dtypes)
```

```python
df['Start Year'] = (
    df['Year(s)']
    .str.extract(r'(\d{4})')[0]
    .astype(int)
)

print(df[['Year(s)', 'Start Year']].head(10))
```

```python
df['Year(s)'] = (
    df['Year(s)']
    .str.replace('–', '-', regex=False)
    .str.strip()
)

print(df['Year(s)'].unique())
```

```python
print("Final Data Types:")
print(df.dtypes)
```

## Data Type Correction

The dataset contained several columns stored as strings even though they represented numerical values.

The `Peak` and `All Time Peak` ranking columns were converted to numeric types. Missing ranking values are represented as `NaN` for numerical processing.

The monetary columns (`Actual gross`, `Adjusted gross`, and `Average gross`) were converted to `float` after removing currency symbols and other non-numeric characters.

The `Rank` and `Shows` columns were already stored as integers and were retained.

The `Year(s)` column was standardized by replacing the en dash with a hyphen and keeping it as a string because some tours span multiple years. A separate `Start Year` integer column was created from the first year of each tour to support numerical analysis.

```python
numeric_columns = [
    'Rank',
    'Peak',
    'All Time Peak',
    'Actual\xa0gross',
    'Adjusted\xa0gross (in 2022 dollars)',
    'Shows',
    'Average gross',
    'Start Year'
]

outlier_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    outlier_summary.append({
        'Column': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df
```

```python
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    
    print(f"\n{column}: {len(outliers)} outliers")
    
    if len(outliers) > 0:
        print(outliers[[column, 'Artist', 'Tour title']].to_string(index=False))
```

## Outlier Detection and Treatment

The Interquartile Range (IQR) method was used to identify potential outliers in the numerical columns.

The analysis identified outliers in `Actual gross`, `Adjusted gross`, `Shows`, `Average gross`, and `Start Year`.

After reviewing the identified records, the outliers were found to represent genuine observations rather than data-entry errors. For example, Taylor Swift's *The Eras Tour* and Beyoncé's *Renaissance World Tour* have exceptionally high gross values because they are among the highest-grossing tours in the dataset. Similarly, Cher's *Living Proof: The Farewell Tour* has a high number of shows and an earlier start year, but these values are valid.

Therefore, no outliers were removed or capped. The identified values were retained because they provide meaningful information about the dataset.
